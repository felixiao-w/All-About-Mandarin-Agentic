# 02 — EDA & Visualisasi Data Bisnis: Mandarin Sentence & Translator Agent

- **Modul Agent:** `sentence_translator`
- **Dataset:** `../data/02_mandarin_sentences.csv`
- **Tujuan:** Menganalisis panjang kalimat, topik percakapan, dan struktur pasangan kalimat Mandarin untuk mendukung konteks penerjemahan pada AI Agent.

## 1. Load Dataset

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

DATA_PATH = Path("../data/02_mandarin_sentences.csv")

try:
    df = pd.read_csv(DATA_PATH)
    print(f"Dataset berhasil dimuat: {df.shape[0]} baris, {df.shape[1]} kolom")
    display(df.head())
except FileNotFoundError:
    print(f"File tidak ditemukan di {DATA_PATH.resolve()}. Unduh dataset lalu simpan ke folder data/.")
    df = pd.DataFrame()
```

## 2. Data Cleaning & Info

```python
if not df.empty:
    print("=== Info Kolom ===")
    df.info()
    print("\n=== Missing Values ===")
    print(df.isnull().sum())
    print(f"\n=== Duplikasi Baris: {df.duplicated().sum()} ===")
    display(df.describe(include="all").T)
else:
    print("Dataset kosong. Lewati tahap cleaning.")
```

## 3. Exploratory Data Analysis & Visualisasi

```python
# TODO: Sesuaikan kolom teks Mandarin dengan skema aktual 02_mandarin_sentences.csv
if not df.empty:
    text_col = df.columns[0]
    lengths = df[text_col].astype(str).str.len()

    plt.figure(figsize=(8, 4))
    sns.histplot(lengths, bins=30, kde=True, color="steelblue")
    plt.title(f"Distribusi Panjang Karakter pada Kolom '{text_col}'")
    plt.xlabel("Panjang Karakter")
    plt.ylabel("Frekuensi")
    plt.tight_layout()
    plt.show()
else:
    print("Muat dataset terlebih dahulu untuk menjalankan visualisasi EDA.")
```
