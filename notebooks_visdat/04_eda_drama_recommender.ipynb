# 04 — EDA & Visualisasi Data Bisnis: Chinese Drama & Movie Recommender Agent

- **Modul Agent:** `drama_recommender`
- **Dataset:** `../data/04_chinese_dramas_movies.csv`
- **Tujuan:** Menganalisis tren genre, rating, tahun rilis, dan popularitas C-Drama/C-Movie sebagai dasar pembobotan rekomendasi AI Agent.

## 1. Load Dataset

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

DATA_PATH = Path("../data/04_chinese_dramas_movies.csv")

try:
    df = pd.read_csv(DATA_PATH)
    print(f"Dataset berhasil dimuat: {df.shape[0]} baris, {df.shape[1]} kolom")
    display(df.head())
except FileNotFoundError:
    print(f"File tidak ditemukan di {DATA_PATH.resolve()}. Unduh dataset lalu simpan ke folder data/.")
    df = pd.DataFrame()
```

## 2. Data Cleaning & Info

```python
if not df.empty:
    print("=== Info Kolom ===")
    df.info()
    print("\n=== Missing Values ===")
    print(df.isnull().sum())
    print(f"\n=== Duplikasi Baris: {df.duplicated().sum()} ===")
    display(df.describe(include="all").T)
else:
    print("Dataset kosong. Lewati tahap cleaning.")
```

## 3. Exploratory Data Analysis & Visualisasi

```python
# TODO: Sesuaikan kolom genre & rating dengan skema aktual 04_chinese_dramas_movies.csv
if not df.empty:
    numeric_cols = df.select_dtypes(include="number").columns.tolist()
    plt.figure(figsize=(8, 4))
    if numeric_cols:
        sns.histplot(df[numeric_cols[0]].dropna(), bins=20, kde=True, color="purple")
        plt.title(f"Distribusi '{numeric_cols[0]}' (Chinese Dramas & Movies)")
        plt.xlabel(numeric_cols[0])
        plt.ylabel("Jumlah Judul")
    else:
        top_vals = df[df.columns[0]].astype(str).value_counts().head(10)
        sns.barplot(x=top_vals.values, y=top_vals.index)
        plt.title("Top 10 Entri Drama & Film")
    plt.tight_layout()
    plt.show()
else:
    print("Muat dataset terlebih dahulu untuk menjalankan visualisasi EDA.")
```
