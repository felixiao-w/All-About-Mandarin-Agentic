# 05 — EDA & Visualisasi Data Bisnis: Chinese Culinary & Menu Agent

- **Modul Agent:** `culinary_menu`
- **Dataset:** `../data/05_chinese_culinary_menu.csv`
- **Tujuan:** Memetakan keragaman masakan daerah Tiongkok, profil rasa, bahan dominan, dan kategori menu untuk rekomendasi kuliner.

## 1. Load Dataset

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

DATA_PATH = Path("../data/05_chinese_culinary_menu.csv")

try:
    df = pd.read_csv(DATA_PATH)
    print(f"Dataset berhasil dimuat: {df.shape[0]} baris, {df.shape[1]} kolom")
    display(df.head())
except FileNotFoundError:
    print(f"File tidak ditemukan di {DATA_PATH.resolve()}. Unduh dataset lalu simpan ke folder data/.")
    df = pd.DataFrame()
```

## 2. Data Cleaning & Info

```python
if not df.empty:
    print("=== Info Kolom ===")
    df.info()
    print("\n=== Missing Values ===")
    print(df.isnull().sum())
    print(f"\n=== Duplikasi Baris: {df.duplicated().sum()} ===")
    display(df.describe(include="all").T)
else:
    print("Dataset kosong. Lewati tahap cleaning.")
```

## 3. Exploratory Data Analysis & Visualisasi

```python
# TODO: Sesuaikan kolom kategori rasa / wilayah asal dengan skema aktual 05_chinese_culinary_menu.csv
if not df.empty:
    target_col = df.columns[0]
    top_values = df[target_col].astype(str).value_counts().head(10)

    plt.figure(figsize=(8, 4))
    sns.barplot(x=top_values.values, y=top_values.index, palette="rocket")
    plt.title(f"Top 10 Distribusi Kolom '{target_col}' (Chinese Culinary Menu)")
    plt.xlabel("Jumlah Menu")
    plt.ylabel(target_col)
    plt.tight_layout()
    plt.show()
else:
    print("Muat dataset terlebih dahulu untuk menjalankan visualisasi EDA.")
```
